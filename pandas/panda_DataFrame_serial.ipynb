## Importing required library 

```python
import pandas as pd 
import numpy as np 
```

```python
df =pd.DataFrame({
    'Name':['Sarah','Rahim',np.nan,'Mim'],
    'Age':[25,np.nan, 22,22],
    'City':['Rajshahi',np.nan,'Khulna','Dhaka']
})
display(df)
```

```python
age=df['Age']
type(age)
print('Printing real age')
display(age)
print('\n
```

```python

count =age.value_counts()
display(count)
type(count)
```

```python
renamed=count.rename_axis('my_value')
print(f'Type for count {type(count)} and type for renamed {type(renamed)}')
display(renamed)
```
