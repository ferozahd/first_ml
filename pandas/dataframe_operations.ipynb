```python
from itertools import groupby

import pandas as pd
import numpy as np
```

# Goal is to understand
 - transformer

```python
df = pd.DataFrame({
    "customer": ["A", "A", "A", "A", "B", "B", "B", "B"],
    "date": pd.to_datetime([
        "2026-08-01", "2026-08-02", "2026-08-03", "2026-08-04",
        "2026-08-01", "2026-08-02", "2026-08-03", "2026-08-04"
    ]),
    "amount": [100, 200, 300, 400, 50, 150, 250, 350]
})
```

```python
df
```

## Let's understand aggregation vs Transformer

```python
a_df=df.groupby("customer")["amount"].mean()
a_df
```

```python
df['customer_mean']=df.groupby("customer")["amount"].transform('mean')
df
```

## Outcome
- groupby aggregation will create report by groupby function
- aggregation will return grouping entire data

### Transform
- It will apply group result to all dataset with same indecis order
