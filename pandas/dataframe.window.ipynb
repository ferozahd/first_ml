## This part I will look for exploring window function in dataframe

```python
import pandas as pd
import numpy as np
```

```python

df = pd.DataFrame({
    "day": ["Day 1", "Day 2", "Day 3", "Day 4"],
    "sales": [10, 20, 30, 40]
})

print(df)
```

```python
df["average_last_2"] = (
    df["sales"]
    .rolling(window=2, min_periods=1)
    .mean()
)

print(df)
```

### What we have learned here
- operation not exactly in same column but
- I can make operation in completely 2 different columns
- After operation I can set result it records

| index | sales | average_last_2 |               math                |               rules                |
|:-----:|:-----:|:--------------:|:---------------------------------:|:----------------------------------:|
|   1   |  10   |       10       | No/ because previous value exists | No because no previous value exist |
|   2   |  20   |       15       |             (10+20)/2             |        index(1)+index(2)/2         |
|   3   |  30   |       25       |             (20+30)/2             |        index(2)+index(3)/2         |
|   4   |  40   |       35       |             (30+40)/2             |        index(3)+index(4)/2         |


```python
df["sum_last_2"] = (
    df["sales"]
    .rolling(window=2, min_periods=1)
    .sum()
)

print(df)
```

```python
df['email_attempt_last_5m']=(
   df.groupby("email", sort=False)
           .rolling(
               window="5min",
               on="time_to_attempt",
               closed="right"
           )["id"]
           .count()
           .astype(int)
           .to_numpy()
)

```

```python
df["_original_row"] = np.arange(len(df))


# --------------------------------------------------
# 2. Sort by email and timestamp
# --------------------------------------------------

test_df = (
    df.sort_values(
        ["email", "time_to_attempt", "id"],
        kind="stable"
    )
    .copy()
)


# --------------------------------------------------
# 3. Calculate the expected correct value
# --------------------------------------------------

test_df["expected_email_attempt_last_5m"] = (
    test_df.groupby("email", sort=False)
           .rolling(
               window="5min",
               on="time_to_attempt",
               closed="right"
           )["id"]
           .count()
           .astype(int)
           .to_numpy()
)


# --------------------------------------------------
# 4. Return to the original CSV order
# --------------------------------------------------

test_df = (
    test_df.sort_values("_original_row")
           .reset_index(drop=True)
)


# --------------------------------------------------
# 5. Compare existing and expected values
# --------------------------------------------------

test_df["is_correct"] = (
    test_df["email_attempt_last_5m"]
    == test_df["expected_email_attempt_last_5m"]
)


# Show summary
total_rows = len(test_df)
correct_rows = test_df["is_correct"].sum()
incorrect_rows = (~test_df["is_correct"]).sum()

print("Total rows:", total_rows)
print("Correct rows:", correct_rows)
print("Incorrect rows:", incorrect_rows)


# --------------------------------------------------
# 6. Show only incorrect records
# --------------------------------------------------

incorrect_records = test_df.loc[
    ~test_df["is_correct"],
    [
        "id",
        "email",
        "time_to_attempt",
        "email_attempt_last_5m",
        "expected_email_attempt_last_5m"
    ]
]

print("\nIncorrect records:")
print(incorrect_records.head(30))
```
