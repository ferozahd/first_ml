# Data Selection and Filtering 

### DataFrame `DataFrame.loc[]` explore

`DataFrame.loc[]` selects or updates rows and columns using labels or conditions.

#### Example
> Select one rows
```python
print(df.loc[1])
```
> Select rows and columns
```python
print(df.loc[0:1, ["name", "age"]])
```
> Select row using condition
```python

print(df.loc[df["city"] == "Dhaka"])
```
> Update values on condition
```python
df.loc[df["city"] == "Dhaka", "city"] = "Bangladesh"
```
>Update one cell by index and column name
```python
df.loc[0, "age"] = 26

```

#### Basic syntax

```python
df.loc[row_selector, column_selector]
```

````python

loc_df = pd.DataFrame({
    "name": ["Sara", "Alex", "John", "Emma"],
    "age": [25, 30, 22, 28],
    "city": ["Dhaka", "London", "Dhaka", "Paris"]
})

print("Original DataFrame:")
print(loc_df)

# Select one row by index label
row = loc_df.loc[1]

print("\nRow at index 1:")
print(row)

# Select specific rows and columns
selected_data = loc_df.loc[1:2, ["name", "age"]]

print("\nSelected rows and columns:")
print(selected_data)

# Select rows using a condition
dhaka_people = loc_df.loc[loc_df["city"] == "Dhaka"]

print("\nPeople from Dhaka:")
print(dhaka_people)

# Select specific columns using a condition
adult_names = loc_df.loc[loc_df["age"] >= 25, ["name", "age"]]

print("\nPeople aged 25 or above:")
print(adult_names)

# Update values using loc
loc_df.loc[loc_df["city"] == "Dhaka", "city"] = "Bangladesh"

print("\nUpdated DataFrame:")
print(loc_df)
````

### Pandas `DataFrame.iloc[]`

>`DataFrame.iloc[]` selects or updates rows and columns using integer positions.

## Example
> Select first row
```python
print(df.iloc[0])
```
>Select rowws from position 0 to 1
```python
print(df.iloc[0:2])
```
> Select specific rows and columns 
```python
print(df.iloc[0:2, 0:2])
```
> Select one specific value
```python
print(df.iloc[1, 2])
```

```python

```
> Update an entire column
```python

```

#### Basic syntax

```python
df.iloc[row_position, column_position]
```

#### Common examples

```python
df.iloc[0]          # First row
df.iloc[-1]         # Last row
df.iloc[:, 0]       # First column
df.iloc[0:3]        # First three rows
df.iloc[0:3, 0:2]   # First three rows and first two columns
```


````python

iloc_df = pd.DataFrame({
    "name": ["Sara", "Alex", "John", "Emma", "David"],
    "age": [25, 30, 22, 28, 35],
    "city": ["Dhaka", "London", "Paris", "Tokyo", "Sydney"],
    "score": [85, 90, 78, 88, 92]
})

print("Original DataFrame:")
print(iloc_df)

# Select the first row
first_row = iloc_df.iloc[0]
print("\nFirst row:")
print(first_row)

# Select the last row
last_row = iloc_df.iloc[-1]
print("\nLast row:")
print(last_row)

# Select the first three rows
first_three_rows = iloc_df.iloc[0:3]
print("\nFirst three rows:")
print(first_three_rows)

# Select the first column
first_column = iloc_df.iloc[:, 0]
print("\nFirst column:")
print(first_column)

# Select multiple columns
selected_columns = iloc_df.iloc[:, [0, 2]]
print("\nSelected columns:")
print(selected_columns)

# Select specific rows and columns
selected_data = iloc_df.iloc[1:4, 0:3]
print("\nSelected rows and columns:")
print(selected_data)

# Select one specific value
selected_value = iloc_df.iloc[1, 2]
print("\nSpecific value:")
print(selected_value)

# Select non-consecutive rows
selected_rows = iloc_df.iloc[[0, 2, 4]]
print("\nNon-consecutive rows:")
print(selected_rows)

# Select non-consecutive rows and columns
selected_rows_columns = iloc_df.iloc[[0, 2, 4], [0, 3]]
print("\nNon-consecutive rows and columns:")
print(selected_rows_columns)

# Update one cell
iloc_df.iloc[0, 1] = 26

# Update multiple cells in one column
iloc_df.iloc[0:3, 3] = [87, 93, 80]

# Update an entire column
iloc_df.iloc[:, 2] = [
    "Bangladesh",
    "United Kingdom",
    "France",
    "Japan",
    "Australia"
]

print("\nUpdated DataFrame:")
print(iloc_df)

# Select all rows except the last row
all_except_last = iloc_df.iloc[:-1]
print("\nAll rows except the last row:")
print(all_except_last)

# Select all columns except the last column
all_except_last_column = iloc_df.iloc[:, :-1]
print("\nAll columns except the last column:")
print(all_except_last_column)

# Select the last two rows and last two columns
last_two_rows_columns = iloc_df.iloc[-2:, -2:]
print("\nLast two rows and last two columns:")
print(last_two_rows_columns)
````

### Pandas `DataFrame.drop_duplicates()`

>`DataFrame.drop_duplicates()` removes duplicate rows from a DataFrame.

#### Example

> Remove completely duplicate rows
```python
all_duplicates_removed = duplicate_df.drop_duplicates()
```
> Remove duplicates based on the "name" column
```python
name_duplicates_removed = duplicate_df.drop_duplicates(
    subset=["name"]
)
```
> Keep the last duplicate occurrence
```python
keep_last_duplicates = duplicate_df.drop_duplicates(
    subset=["name"],
    keep="last"
)
```
> Remove all rows containing duplicate names
```python
remove_all_duplicates = duplicate_df.drop_duplicates(
    subset=["name"],
    keep=False
)
```
> Remove duplicates and reset the index
```python
reset_index_df = duplicate_df.drop_duplicates(
    ignore_index=True
)
```
> Update the original DataFrame

```python 
duplicate_df.drop_duplicates(
    inplace=True,
    ignore_index=True
)
```

## Basic syntax

```python
duplicate_df.drop_duplicates(
    subset=None,
    keep="first",
    inplace=False,
    ignore_index=False
)
```

## Important parameters

* `subset`: Columns used to identify duplicate rows.
* `keep="first"`: Keeps the first duplicate occurrence.
* `keep="last"`: Keeps the last duplicate occurrence.
* `keep=False`: Removes every duplicated occurrence.
* `inplace=True`: Updates the original DataFrame.
* `ignore_index=True`: Resets the index after removing duplicates.

## Common examples

```python
duplicate_df.drop_duplicates()

duplicate_df.drop_duplicates(subset=["name"])

duplicate_df.drop_duplicates(subset=["name", "city"])

duplicate_df.drop_duplicates(subset=["name"], keep="last")

duplicate_df.drop_duplicates(subset=["name"], keep=False)

duplicate_df.drop_duplicates(ignore_index=True)

duplicate_df.drop_duplicates(inplace=True)
```


````python

duplicate_df = pd.DataFrame({
    "name": ["Sara", "Alex", "Sara", "John", "Alex", "Emma"],
    "age": [25, 30, 25, 22, 30, 28],
    "city": ["Dhaka", "London", "Dhaka", "Paris", "London", "Dhaka"]
})

print("Original DataFrame:")
print(duplicate_df)

# Remove completely duplicate rows
all_duplicates_removed = duplicate_df.drop_duplicates()

print("\nAfter removing completely duplicate rows:")
print(all_duplicates_removed)

# Remove duplicates based only on the "name" column
name_duplicates_removed = duplicate_df.drop_duplicates(
    subset=["name"]
)

print("\nAfter removing duplicate names:")
print(name_duplicates_removed)

# Keep the last duplicate occurrence
keep_last_duplicates = duplicate_df.drop_duplicates(
    subset=["name"],
    keep="last"
)

print("\nKeep the last occurrence:")
print(keep_last_duplicates)

# Remove every row that has a duplicate name
remove_all_duplicates = duplicate_df.drop_duplicates(
    subset=["name"],
    keep=False
)

print("\nRemove all duplicated names:")
print(remove_all_duplicates)

# Remove duplicates and reset the index
reset_index_df = duplicate_df.drop_duplicates(
    ignore_index=True
)

print("\nRemove duplicates and reset index:")
print(reset_index_df)

# Modify the original DataFrame directly
duplicate_df.drop_duplicates(
    inplace=True,
    ignore_index=True
)

print("\nUpdated original DataFrame:")
print(duplicate_df)
````

$$Common aggregation methods$$
#### Common `groupby().agg()` methods

`first`

Returns the first non-null value from each group.

`last`

Returns the last non-null value from each group.

`sum`

Returns the total of all values in each group.

`mean`

Returns the average value of each group.

`median`

Returns the median value of each group.

`min`

Returns the minimum value from each group.

`max`

Returns the maximum value from each group.

`count`

Returns the number of non-null values in each group.

`size`

Returns the total number of rows in each group, including null values.

`nunique`

Returns the number of unique values in each group.

`prod`

Returns the product of all values in each group.

`std`

Returns the standard deviation of each group.

`var`

Returns the variance of each group.

`sem`

Returns the standard error of the mean for each group.

`quantile`

Returns a specified quantile value from each group.

`any`

Returns `True` if at least one value in the group is true.

`all`

Returns `True` if all values in the group are true.

`idxmin`

Returns the index of the minimum value in each group.

`idxmax`

Returns the index of the maximum value in each group.

`skew`

Returns the skewness of values in each group.

`cov`

Returns the covariance between columns within each group.


````python

groupby_df = pd.DataFrame({
    "department": ["IT", "IT", "HR", "HR", "Sales", "Sales"],
    "employee": ["Sara", "Alex", "John", "Emma", "David", "Sophia"],
    "city": ["Dhaka", "London", "Dhaka", "London", "Dhaka", "London"],
    "salary": [50000, 60000, 45000, 48000, 55000, 65000],
    "sales": [20, 25, 10, 15, 30, 35]
})

print("Original DataFrame:")
print(groupby_df)

# Total salary for each department
department_salary = groupby_df.groupby("department")["salary"].sum()

print("\nTotal salary by department:")
print(department_salary)

# Average salary for each department
average_salary = groupby_df.groupby("department")["salary"].mean()

print("\nAverage salary by department:")
print(average_salary)

# Number of employees in each department
employee_count = groupby_df.groupby("department").size()

print("\nEmployee count by department:")
print(employee_count)

# Group using multiple columns
city_department_salary = (
    groupby_df
    .groupby(["city", "department"])["salary"]
    .sum()
)

print("\nSalary grouped by city and department:")
print(city_department_salary)

# Apply multiple calculations
department_summary = (
    groupby_df
    .groupby("department")
    .agg(
        total_salary=("salary", "sum"),
        average_salary=("salary", "mean"),
        highest_salary=("salary", "max"),
        total_sales=("sales", "sum")
    )
    .reset_index()
)

print("\nDepartment summary:")
print(department_summary)

# Keep department as a normal column
grouped_with_column = (
    groupby_df
    .groupby("department", as_index=False)["salary"]
    .sum()
)

print("\nGrouped result with a normal index:")
print(grouped_with_column)

# Add department average beside every employee
groupby_df["department_average_salary"] = (
    groupby_df
    .groupby("department")["salary"]
    .transform("mean")
)

print("\nDataFrame with department average salary:")
print(groupby_df)

# Keep groups whose total sales are greater than 40
filtered_groups = groupby_df.groupby("department").filter(
    lambda group: group["sales"].sum() > 40
)

print("\nDepartments with total sales greater than 40:")
print(filtered_groups)
````

### Pandas `DataFrame.sort_values()`

> Brief definition

>`DataFrame.sort_values()` sorts DataFrame rows by the values of one or more columns.

**Basic syntax**

```python

    by="column_name",

```

**Sort one column in ascending order**

```python

```

**Sort one column in descending order**

```python

    by="price",
    ascending=False
)
```

**Sort using multiple columns**

```python

    by=["category", "price"]
)
```

**Use different sorting orders**

```python
sort_values_df.sort_values(
    by=["category", "price"],
    ascending=[True, False]
)
```

**Place missing values first**

```python
sort_values_df.sort_values(
    by="price",
    na_position="first"
)
```

**Place missing values last**

```python
sort_values_df.sort_values(
    by="price",
    na_position="last"
)
```

**Reset the index after sorting**

```python
sort_values_df.sort_values(
    by="price",
    ignore_index=True
)
```

**Update the original DataFrame**

```python
sort_values_df.sort_values(
    by="price",
    inplace=True
)
```

### Important parameters

**`by`**

>Specifies the column or columns used for sorting.

**`ascending=True`**

Sorts values from smallest to largest.

**`ascending=False`**

Sorts values from largest to smallest.

**`na_position="first"`**

Places missing values at the beginning.

**`na_position="last"`**

Places missing values at the end.

**`ignore_index=True`**

Creates a new sequential index after sorting.

**`inplace=True`**

Updates the original DataFrame directly.


````python

sort_values_df = pd.DataFrame({
    "item": ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphone"],
    "category": [
        "Computer",
        "Accessory",
        "Accessory",
        "Computer",
        "Accessory"
    ],
    "price": [800, 20, 50, 300, 80],
    "quantity": [5, 20, 10, 7, 12]
})

print("Original DataFrame:")
print(sort_values_df)

# Sort price in ascending order
price_ascending = sort_values_df.sort_values(
    by="price"
)

print("\nPrice in ascending order:")
print(price_ascending)

# Sort price in descending order
price_descending = sort_values_df.sort_values(
    by="price",
    ascending=False
)

print("\nPrice in descending order:")
print(price_descending)

# Sort using category and price
multiple_column_sort = sort_values_df.sort_values(
    by=["category", "price"],
    ascending=[True, False]
)

print("\nSort by category ascending and price descending:")
print(multiple_column_sort)

# Sort quantity and reset the index
quantity_sorted = sort_values_df.sort_values(
    by="quantity",
    ascending=False,
    ignore_index=True
)

print("\nQuantity sorted with a new index:")
print(quantity_sorted)

# Update the original DataFrame
sort_values_df.sort_values(
    by="price",
    ascending=True,
    ignore_index=True,
    inplace=True
)

print("\nUpdated original DataFrame:")
print(sort_values_df)
````

### Pandas `DataFrame.apply()`

>Brief definition

`DataFrame.apply()` applies a function to each column or each row of a DataFrame.

It does not normally apply the function to every individual cell. For element-wise operations, use `DataFrame.map()`.

**Basic syntax**

```python
apply_df.apply(function)
```

**Apply to each column**

`axis=0` applies the function to each column.

```python
apply_df.apply(function, axis=0)
```

**Apply to each row**

`axis=1` applies the function to each row.

```python
apply_df.apply(function, axis=1)
```

**Find the maximum of each numeric column**

```python
apply_df[["price", "quantity"]].apply(max)
```

**Find the sum of each numeric column**

```python
apply_df[["price", "quantity"]].apply(sum)
```

**Use a lambda function on columns**

```python
apply_df[["price", "quantity"]].apply(
    lambda column: column.mean()
)
```

## Create a value using multiple columns

```python
apply_df["total"] = apply_df.apply(
    lambda row: row["price"] * row["quantity"],
    axis=1
)
```

**Create a category using row values**

```python
apply_df["price_level"] = apply_df.apply(
    lambda row: "Expensive" if row["price"] >= 100 else "Affordable",
    axis=1
)
```

**Apply a named function**

```python
def calculate_total(row):
    return row["price"] * row["quantity"]
```

```python
apply_df["total"] = apply_df.apply(
    calculate_total,
    axis=1
)
```

**Return multiple columns**

```python
def calculate_details(row):
    return pd.Series({
        "total": row["price"] * row["quantity"],
        "available": row["quantity"] > 0
    })
```

```python
details = apply_df.apply(
    calculate_details,
    axis=1
)
```

### Important parameters

`func`

The function that will be applied.

`axis=0`

Applies the function to each column. This is the default.

`axis=1`

Applies the function to each row.

`args`

Passes additional positional arguments to the function.

`result_type="expand"`

Expands list-like row results into separate columns.

`apply()` versus `map()`

Use `apply()` when the function needs an entire row or column.

```python
apply_df.apply(function, axis=1)
```

Use `map()` when the function should process every individual cell.

```python
apply_df.map(function)
```


````python


apply_df = pd.DataFrame({
    "item": ["Laptop", "Mouse", "Keyboard", "Monitor"],
    "price": [800, 20, 50, 300],
    "quantity": [2, 10, 5, 3]
})

print("Original DataFrame:")
print(apply_df)

# Apply sum to each selected column
column_totals = apply_df[["price", "quantity"]].apply(
    sum,
    axis=0
)

print("\nColumn totals:")
print(column_totals)

# Apply mean to each selected column
column_averages = apply_df[["price", "quantity"]].apply(
    lambda column: column.mean(),
    axis=0
)

print("\nColumn averages:")
print(column_averages)

# Calculate total price using each row
apply_df["total_price"] = apply_df.apply(
    lambda row: row["price"] * row["quantity"],
    axis=1
)

# Create a category using each row
apply_df["price_level"] = apply_df.apply(
    lambda row: "Expensive"
    if row["price"] >= 100
    else "Affordable",
    axis=1
)

# Apply a named function
def calculate_discount(row):
    if row["quantity"] >= 5:
        return row["total_price"] * 0.10
    return 0


apply_df["discount"] = apply_df.apply(
    calculate_discount,
    axis=1
)

# Calculate final price
apply_df["final_price"] = apply_df.apply(
    lambda row: row["total_price"] - row["discount"],
    axis=1
)

print("\nUpdated DataFrame:")
print(apply_df)
````
