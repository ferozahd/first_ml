```python
import tensorflow as tf
import numpy as np
from tensorflow.keras.preprocessing import image
```

## Load the trained model file here


```python
best_model = tf.keras.models.load_model("cat_dog_classifier.keras")
```

```python

def predict_cat_dog(img_path):
    img = image.load_img(img_path, target_size=(224, 224))
    img_array = image.img_to_array(img)
    img_array = img_array / 255.0
    img_array = np.expand_dims(img_array, axis=0)

    prediction = best_model.predict(img_array)[0][0]
    
    if prediction >= 0.5:
        print("Prediction: Dog")
        print("Confidence:", prediction)
    else:
        print("Prediction: Cat")
        print("Confidence:", 1 - prediction)
```

```python
predict_cat_dog(img_path="../data/cat_dog/test/cats/cat.9993.jpg")
```
