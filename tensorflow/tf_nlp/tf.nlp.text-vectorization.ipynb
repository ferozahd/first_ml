### Important note
`max_tokens=20` Means the layer will keep only the top 20 most common words.<br>
`output_mode="int"` Means every word will become an integer number.

<div style="border:1px solid #999; padding:10px; border-radius:6px; background-color:#A00000">
 `output_sequence_length=6` Means every sentence will have exactly 6 numbers.<br>
 Example:
 <p>"I love cats"</p>
<br>
output:
[3, 6, 8, 0, 0, 0]
</div>

```python
import tensorflow as tf
from tensorflow.keras.layers import TextVectorization
# 1. Some sample text data
texts = [
    "I love cats",
    "I love dogs",
    "Cats are cute",
    "Dogs are loyal",
    "I hate spam messages",
    "Spam messages are annoying"
]

# 2. Create TextVectorization layer
vectorizer =  TextVectorization(
    max_tokens=20,              # maximum vocabulary size
    output_mode="int",          # convert words to integer numbers
    output_sequence_length=6     # every sentence length will be 6
)

# 3. Learn vocabulary from the text data
vectorizer.adapt(texts)

# 4. See vocabulary
vocab = vectorizer.get_vocabulary()
print("Vocabulary:")
print(vocab)

# 5. Test with a new sentence
sample_text = tf.constant(["I love cute cats"])

vectorized_text = vectorizer(sample_text)

print("\nOriginal text:")
print(sample_text.numpy())

print("\nVectorized text:")
print(vectorized_text.numpy())

print("Original text for 'I love tiger'")
print(vectorizer(tf.constant(["I love tiger"])).numpy() )
```
