```python
import spacy

nlp = spacy.load("en_core_web_sm")
text = "I would like to take two units of another items is burger, a burger, a hot coffiee and 2 teas please"

doc= nlp(text)

for token in doc:
    print(token.text ,"->", token.dep_ ,"->", token.head.text)


def extract_quantity_and_anchor(doc):
    quantity = None
    item = None


    for token in doc:
        if token.like_num:
            quantity=token.text

        if token.pos_ == "NOUN":
            item = token.text
    return quantity, item


extract_quantity_and_anchor(doc)
```
