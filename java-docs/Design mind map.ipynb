# Phase 1: Foundation — Visual Thinking Mindset

Before learning charts, learn **what problem you are solving**.

Every visual should answer one of these questions:

### 1. Who/What?

Identity, description, components, input data , user , responsible webhook 

Example:

* What is a company?
* What are the parts of a machine?

Visuals:

* Mind map
* Concept map
* Tree diagram
* Organizational chart

### 2. How much?

Quantitative comparison

Example:

* Sales by country
* Market share

Visuals:

* Bar chart
* Pie chart
* Scatter plot
* Histogram

### 3. Where?

Position, location, relationship , Computer, processor , Database , RAM , Router , Server 

Example:

* Customer journey
* Business ecosystem

Visuals:

* Maps
* Matrix
* 2×2 framework

### 4. When?

Time and sequence, Specific occation, when logic satisfy, in schadule 

Example:

* Project timeline
* Historical events

Visuals:

* Timeline
* Gantt chart
* Roadmap

### 5. How?

Process and movement, Algorithm , flow-chart 

Example:

* How does a product launch happen?

Visuals:

* Flowchart
* Process map
* Funnel

### 6. Why?

Cause and reasoning, Policy, rules, Constrain , Features, Domain needs  

Example:

* Why did sales decrease?

Visuals:

* Fishbone diagram
* Cause-effect tree
* Logic tree

This is the foundation.

# Phase 2: Visual Grammar (Your "Alphabet")

Think of visuals like a language.

Before making complex diagrams, master basic symbols.

## Objects

Represent things:

* Rectangle = object/system
* Circle = idea/category
* Person icon = people
* Document icon = information
* Database icon = storage

## Relationships

Show connection:

* Line = relationship
* Arrow = direction
* Thick line = strong relationship
* Dotted line = weak/optional

## Actions

Show change:

* Increase ↑
* Decrease ↓
* Movement →
* Cycle ↻
* Conflict ⚡

## Emphasis

Show importance:

* Size
* Color
* Position
* Boldness

Think of this as learning vocabulary.

# Phase 3: Frameworks (The Thinking Structures)

This is where your "growth/reduction, timeline, statistical, circular, tree" belongs.

Frameworks are not drawings. They are **ways to organize thinking**.

## 1. Comparison Frameworks

Question:
"What is different?"

Examples:

* Pros vs Cons
* Before vs After
* A vs B

Visuals:

* Table
* 2×2 matrix
* Comparison chart

---

## 2. Change Frameworks

Question:
"What is changing?"

Examples:

Growth:

```
Sales
 |
 |        *
 |     *
 |  *
 |____________
```

Reduction:

```
Cost
 |
 | *
 |    *
 |       *
 |____________
```

Visuals:

* Line chart
* Trend chart
* Waterfall chart

---

## 3. Time Frameworks

Question:
"What happens over time?"

Visuals:

* Timeline
* Roadmap
* Gantt chart

---

## 4. Hierarchy Frameworks

Question:
"How is something structured?"

Visuals:

Tree:

```
Company
 |
 |-- Marketing
 |-- Sales
 |-- Finance
```

Examples:

* Organization
* Classification
* Mind map

---

## 5. Flow Frameworks

Question:
"How does something move?"

Visuals:

```
Input → Process → Output
```

Examples:

* Customer journey
* Supply chain
* Algorithm

---

## 6. Circular Frameworks

Question:
"What repeats?"

Visuals:

Cycle:

```
Plan
 ↓
Do
 ↓
Check
 ↓
Improve
```

Examples:

* Business cycle
* Learning cycle

---

## 7. Statistical Frameworks

Question:
"What does the data say?"

Visuals:

* Bar chart
* Pie chart
* Scatter plot
* Distribution

# Phase 6: Storytelling With Visuals

This is where experts separate themselves.

A good visual story usually follows:

## 1. Context

Where are we?

↓

## 2. Problem

What is wrong?

↓

## 3. Insight

What did we discover?

↓

## 4. Action

What should we do?

Example:

Bad:

"Here is a sales chart."

Good:

"Sales increased 30% after changing pricing strategy."

The visual supports the story.

# Level 1: Business View (Non-technical people)

Question:

**"Why does this system exist?"**

Audience:

* CEO
* Sales
* Accounts
* Banking department
* Product owners

Visual framework:

## Ecosystem Map

Example:

```
              Customers
                  |
                  |
Bank ---- Payment System ---- Business
                  |
                  |
             Government API
```

Shows:

* Who interacts with the system?
* What value does it provide?
* What external systems exist?

This is your "Who/What" framework.

---

# Level 2: Capability View

Question:

**"What can this system do?"**

Example:

```
                 Banking Platform

        ----------------------------
        |                          |
   Loan Management          Payment System
        |
   Customer Management
        |
   Reporting
```

This is a hierarchy/tree framework.

Useful for:

* Business discussion
* Product planning
* Feature expansion

---

# Level 3: System Context View

Question:

**"What systems communicate with this system?"**

Example:

```
              Mobile App
                  |
                  |
Customer --> Core System <-- Bank API
                  |
                  |
             Notification
```

Framework:

Network / relationship map.

This helps debugging because you see dependencies.

---

# Level 4: Container Architecture View

Now technical people enter.

Question:

**"What are the major technical parts?"**

Example:

```
              Frontend

                 |
                 |

          API Gateway

          /          \

 User Service     Payment Service

          \          /

             Database
```

Frameworks used:

* Component hierarchy
* Flow
* Dependency map

This is where microservices, modules, databases appear.

---

# Level 5: Component View

Question:

**"Inside each service, how is it organized?"**

Example:

```
Payment Service

      |
      |
 Controller

      |
      |
 Business Logic

      |
      |
 Repository

      |
      |
 Database
```

Framework:

Tree + flow.

---

# Level 6: Data Flow View

Question:

**"How does information move?"**

This is extremely important.

Example:

```
Customer
   |
   ↓
API
   |
   ↓
Validation
   |
   ↓
Payment Service
   |
   ↓
Bank
```

Framework:

Flow diagram.

Useful for:

* Debugging
* Performance
* Security

# Level 8: Decision & Evolution View

This is where architecture becomes mature.

Question:

**"Why did we choose this design?"**

Example:

```
Problem:
10 million users

      ↓

Decision:
Separate payment service

      ↓

Benefit:
Independent scaling
```

Framework:

Cause → Decision → Consequence.
