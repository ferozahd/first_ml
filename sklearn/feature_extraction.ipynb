```python
from sklearn.feature_extraction.text import (
    CountVectorizer,
    TfidfVectorizer
)
```

## Count Vectorizer 
- Its counting work's frequency such as number of time a words come in dataset 

```python
texts =["I love machine learning","I love deep learning"]
vectorizer =CountVectorizer()
x =vectorizer.fit_transform(texts)
print(vectorizer.get_feature_names_out())
print(x.toarray())
```

## TfidfVectorizer 
`TfidfVectorizer` convert text into numbers by measuring word importance 
A word is important if:
- It appears often in one document 
- but does not appear in every document 
  
**Example**
>I love machine learning 
>I love deep learning 
here `I love learning` present in both sentences but<br>
`machine` & `deep` is not common in both so it bear more importance 

```python
texts =["I love machine learning","I love deep learning"]
vectorizer =TfidfVectorizer()
x =vectorizer.fit_transform(texts)

print(vectorizer.get_feature_names_out())
print(x.toarray())
```
