```python
import pandas as pd 
from sklearn.preprocessing import(
    StandardScaler,
    MinMaxScaler, 
    RobustScaler,
    OneHotEncoder,
    OrdinalEncoder,
    LabelEncoder
)
```

```python
df=pd.DataFrame({
    "age":[18,25,40,60,100],
    "salary":[10000,25000,40000,80000,100000],
    "city":["Dhaka" ,"Chittagong","Dhaka","Sylet","Rajshahi"],
    "education":["SSC","HSC","Bachelor","Master","PhD"],
    "target":["yes","no","yes","no","yes"]
})
```

Example for `StandardScaler`

Use case : 
- Liner regression 
- Logistic regression 
- SVM 
- KNN
- PCA
- Neural Network 


>When to use `StandardScaler` vs `MinMaxScaler`
- When we need something scale in between 0 to 1 as percentage theory 
  - [&cross;] StandardScaler 
  - [&check;] MinMaxScaler


```python
scaler =StandardScaler()
scaled = scaler.fit_transform(df[['age','salary']])
print(scaled)
```

Example for `MinMaxScaler`

Use case :
- Neural Netwrok 
- KNN
-  distance-based models 
-  image/pixel scaling 

>Problem: __Sensitive to outliers__

How its works:
- It working by using most minimum value and most maximum value
- if min =0 and max=100 , any number betwen this working like percentage 
- [0,20,100]
  - 0 &rarr; 0
  - 2 &rarr; 0.2
  - 0 &rarr; 1
-  if [0,40,80,200]
   -  0 &rarr; 0
   -  20 &rarr; 0.1
   -  80 &rarr; 0.4
   -  200 &rarr; 1
> That's why it sensitive with outlier , one big number can chage whole picture 

```python
scaler = MinMaxScaler()
scaled = scaler.fit_transform(df[['age','salary']])
scaled
```

Example for `RobustScaler`

Use case : 
- numerical data with outliers
- No fixed range for input
- No fixed range for output also 
- ### Model 
  - Liner model 
  - SVM 
  - KNN 
  - dataset with extreme values 

How it works :
- It divided range in 2 parts possitive and negative and middle is 0
- example : 1,2,3,4,5 => -2,-1,0,1,2 
  - 1 &rarr; -2
  - 2 &rarr; -1
  - 3 &rarr; 0 _(exact middle)_ 
  - 4 &rarr; 1
  - 5 &rarr; 2

```python
scaler = RobustScaler()
scaled =scaler.fit_transform(df[['age','salary']])
scaled
```

Example for `OneHotEncoder` 

Use case : 
- Input range is categorical 
  - Example : (Dhaka , Chittagong, Sylhet , Rajshahi )
  - Output(column) : (city_Dhaka, city_Chittagong, city_sylhet, city_Rajshahi )
  - Output(column-value): 0 and 1
- Model (Mostly depend on use case)

```python
encoder =OneHotEncoder(sparse_output=False)
encoded = encoder.fit_transform(df[['city']])
print(encoder.get_feature_names_out())
encoded
```

Example for `OrdinalEncoder`

Use case : 
- Input range : Ordered categorical text
- output : numeric integers as float 
- When to use :
  - categories have meaningful order 
- Good data example
  - education level , ssc , hsc , bachelor , master , PhD
  - level : low , medium , large 
- May be after applying `OrdinalEncoder` we need another level of encoding for solving outliner 

```python
order=[["SSC","HSC","Bachelor","Master","PhD"]]
encoder = OrdinalEncoder(categories=order)
encoded = encoder.fit_transform(df[['education']])

encoded
```

Example for `LabelEncoder`

Use case :
- Input to number for most of model training prepocessing text/label data 
- Though it's giving each label with different number , so input parameter could create confusion 
  - given dataset could learn mistakenly 
  - if Rajshahi =1, Dhaka =2 then model could think Dhaka is bigger than Rajshahi 
  - We should use it wthout any hesitation for output case , 
  

```python
encoder = LabelEncoder()
encoded = encoder.fit_transform(df['target'])
encoded_city = encoder.fit_transform(df['city'])
print(encoded)
print(f'Encoded city {encoded_city}')
```
