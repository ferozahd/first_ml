# Exploring ROC Curve
```python 
from sklearn.metrics import roc_curve
```
This method takes 2 things $\text{y\_true}$ and $\text{y\_score}$
& retruns $fpr$ array,$tpr$ array and $threshold$ array 

This $fpr,tpr\text{ and } threshold$ array sequence is very important ,\
$threshold$ sequence said that each threshold hat will be the $tpr\text{ fpr}$

example :
```text 
FPR: [0.  0.  0.5 0.5 1. ]
TPR: [0.  0.5 0.5 1.  1. ]
Thresholds: [ inf 0.8  0.4  0.35 0.1 ]
```
here for Threshold _inf_ we will get $fpr = 0$ and $tpr = 0$ also\
similarly for Threshold _0.8_ we will get $fpr =0$ and $tpr = 0.5$

**Formula:**

- Ture positive ratio , total actual positive by total actual positive and mistakenly positive from actal negative$$\text{TPR}=\frac{TP}{TP + FN}$$

- Miskatenly negative ratio , total mistakenly positive by total actual negative and mistakenly negative from actal negative$$\text{FPR} = \frac{FP}{TN + FP}$$

````python
from typing import Optional
import numpy as np
from numpy.typing import ArrayLike, NDArray
from sklearn.metrics import roc_curve,confusion_matrix
````

````python
def basic_validation(y_true:np.array , y_score:np.array):
       # Basic validation
    if y_true.ndim != 1 or y_score.ndim != 1:
        raise ValueError("y_true and y_score must be one-dimensional.")

    if len(y_true) != len(y_score):
        raise ValueError(
            "y_true and y_score must have the same length."
        )

    if len(y_true) == 0:
        raise ValueError("Input arrays cannot be empty.")

    if not np.all(np.isfinite(y_score)):
        raise ValueError(
            "y_score cannot contain NaN or infinite values."
        )
````

````python
 
def custom_roc_curve(y_true, y_score,drop_intermediate =True):
    y_true = np.asarray(y_true)
    y_score = np.asarray(y_score, dtype=float)

    basic_validation(y_true, y_score)
    # Unique scores used as thresholds, highest first
    thresholds = np.sort(np.unique(y_score))[::-1]

    # Initial threshold: predict no samples as positive
    thresholds = np.r_[np.inf, thresholds]

    tprs=[]
    fprs=[]
    # Confusion matrix 
    for threshold in thresholds:
        y_pred=np.where(y_score>=threshold,1,0).astype(np.int8)

        tp = np.sum((y_true == 1) & (y_pred == 1))
        fp = np.sum((y_true == 0) & (y_pred == 1))

        total_positive = np.sum(y_true == 1)
        total_negative = np.sum(y_true == 0)

        tpr = tp / total_positive if total_positive > 0 else 0.0
        fpr = fp / total_negative if total_negative > 0 else 0.0

        tprs.append(tpr)
        fprs.append(fpr)

    

    return [float(fpr) for fpr in fprs], [float(tpr) for tpr in tprs], thresholds
````

````python
y_true = np.array([0, 0, 1, 1])

y_score = np.array([0.10, 0.40, 0.35,  0.80])

fpr, tpr, thresholds = custom_roc_curve(
    y_true,
    y_score,
    drop_intermediate=False,
)
print("____________FROM CUSTOM_________________")
print("FPR:", fpr)
print("TPR:", tpr)
print("Thresholds:", thresholds)


fpr, tpr, thresholds = roc_curve(
    y_true,
    y_score,
    drop_intermediate=False,
)
print("\n\n____________FROM BUILTI_________________")
print("FPR:", fpr)
print("TPR:", tpr)
print("Thresholds:", thresholds)

````
