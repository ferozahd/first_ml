## Mean Absolute Percentage Error

MAPE is widely used across business, forecasting, and regression modeling. It measures evaluation errors relative to the actual true magnitudes, presenting the model's error as a scale-free percentage.
### Core components 
- True Value ($y_i$): The actual real-world observation for sample $i$, where $y_i \neq 0$ (the metric is undefined if a true value is zero).
- Prediction ($\hat{y}_i$): The value generated by the regression model for sample $i$.
- Absolute Relative Loss (Percentage Error component): The absolute magnitude of the residual normalized against the true value:

$$\left\vert{} \frac{y_i - \hat{y}_i}{y_i} \right\vert{}$$

### The Real Mathematical Formula
In statistical modeling, MAPE is defined as the arithmetic mean of all individual absolute relative errors, scaled by $100$ to represent a percentage value.

For a dataset containing $n$ samples, the formal mathematical definition is:
$$\text{MAPE} = \frac{100\%}{n} \sum_{i=1}^{n} \left\vert{} \frac{y_i - \hat{y}_i}{y_i} \right\vert{}$$


### Step-by-Step Walkthrough with Your Dataset

Let's evaluate the metric using your dataset ($n = 10$).

Actual values ($\mathbf{y}$): $[45, 50, 55, 62, 65, 71, 78, 83, 85, 92]$

Predicted values ($\mathbf{\hat{y}}$): $[45.09, 50.32, 55.54, 60.76, 65.99, 71.21, 76.44, 81.66, 86.88, 92.11]$

### Step A: Calculate individual absolute relative losses

We compute the component $\left\vert{} \frac{y_i - \hat{y}_i}{y_i} \right\vert{}$ for each pair:
- $i=1: \left\vert{} \frac{45 - 45.09}{45} \right\vert{} = \left\vert{} \frac{-0.09}{45} \right\vert{} = \mathbf{0.002000}$
- $i=2: \left\vert{} \frac{50 - 50.32}{50} \right\vert{} = \left\vert{} \frac{-0.32}{50} \right\vert{} = \mathbf{0.006400}$
- $i=3: \left\vert{} \frac{55 - 55.54}{55} \right\vert{} = \left\vert{} \frac{-0.54}{55} \right\vert{} \approx \mathbf{0.009818}$
- $i=4: \left\vert{} \frac{62 - 60.76}{62} \right\vert{} = \left\vert{} \frac{1.24}{62} \right\vert{} = \mathbf{0.020000}$
- $i=5: \left\vert{} \frac{65 - 65.99}{65} \right\vert{} = \left\vert{} \frac{-0.99}{65} \right\vert{} \approx \mathbf{0.015231}$
- $i=6: \left\vert{} \frac{71 - 71.21}{71} \right\vert{} = \left\vert{} \frac{-0.21}{71} \right\vert{} \approx \mathbf{0.002958}$
- $i=7: \left\vert{} \frac{78 - 76.44}{78} \right\vert{} = \left\vert{} \frac{1.56}{78} \right\vert{} = \mathbf{0.020000}$
- $i=8: \left\vert{} \frac{83 - 81.66}{83} \right\vert{} = \left\vert{} \frac{1.34}{83} \right\vert{} \approx \mathbf{0.016145}$
- $i=9: \left\vert{} \frac{85 - 86.88}{85} \right\vert{} = \left\vert{} \frac{-1.88}{85} \right\vert{} \approx \mathbf{0.022118}$
- $i=10:\left\vert{} \frac{92 - 92.11}{92} \right\vert{} = \left\vert{} \frac{-0.11}{92} \right\vert{} \approx \mathbf{0.001196}$

### Step B: Sum the components and take the average
$$\sum_{i=1}^{10} \left\vert{} \frac{y_i - \hat{y}_i}{y_i} \right\vert{} \approx 0.115866$$
$$\text{Average Relative Error} = \frac{0.115866}{10} = \mathbf{0.0115866}$$

### Step C: Scale to Percentage

$$\text{MAPE} = 0.0115866 \times 100\% \approx \mathbf{1.1587\%}$$


```python
def custom_mape(y_true, y_pred):
    """
    Custom implementation of Mean Absolute Percentage Error (MAPE)
    mirroring the raw mathematical average of absolute relative losses.
    """
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    
    # Calculate absolute relative losses: |(y - y_hat) / y|
    absolute_relative_losses = np.abs((y_true - y_pred) / y_true)
    
    # Return mean scaled to percentage format
    return np.mean(absolute_relative_losses) * 100
```

```python
import numpy as np
from sklearn.metrics import mean_absolute_percentage_error



# -------------------------------------------------------------
# Execution with your dataset
# -------------------------------------------------------------
actual_values = [45, 50, 55, 62, 65, 71, 78, 83, 85, 92]
predicted_values = [45.09, 50.32, 55.54, 60.76, 65.99, 71.21, 76.44, 81.66, 86.88, 92.11]

# 1. Built-in Scikit-Learn method 
# Note: sklearn returns the fractional value (0.011587), so we multiply by 100 for true percentage format.
builtin_result = mean_absolute_percentage_error(actual_values, predicted_values) * 100

# 2. Custom Method
custom_result = custom_mape(actual_values, predicted_values)

print(f"MAPE (Scikit-Learn Built-in): {builtin_result:.4f}%")
print(f"MAPE (Custom Implementation):  {custom_result:.4f}%")
```
