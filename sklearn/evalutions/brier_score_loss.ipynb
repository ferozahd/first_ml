```python
import numpy as np
from sklearn.metrics import brier_score_loss

def custom_brier_score(y_true, y_prob):
    """
    Custom implementation of Brier Score Loss
    mirroring the mean squared probability error formula.
    """
    y_true = np.array(y_true)
    y_prob = np.array(y_prob)
    
    # Calculate squared differences between actual binary labels and probabilities
    squared_residuals = (y_true - y_prob) ** 2
    
    # Return the arithmetic mean
    return np.mean(squared_residuals)

```

# Execution with our tracked walkthrough data

```python

true_labels = [1, 0, 1, 0, 0]
model_probabilities = [0.90, 0.10, 0.40, 0.80, 0.20]

# 1. Built-in Scikit-Learn method
builtin_result = brier_score_loss(true_labels, model_probabilities)

# 2. Custom Method
custom_result = custom_brier_score(true_labels, model_probabilities)

print(f"Brier Score Loss (Scikit-Learn Built-in): {builtin_result:.4f}")
print(f"Brier Score Loss (Custom Implementation):  {custom_result:.4f}")
```
