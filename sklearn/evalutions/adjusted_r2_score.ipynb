# Adjusted R-Squared ($R^2$) Score

In regression analysis, **Adjusted R-Squared** is a modified version of the standard R-Squared ($R^2$) metric that accounts for the number of predictors (independent variables) in a model. 

---

## The Limitation of Standard $R^2$

Standard $R^2$ measures the proportion of variance in the dependent variable that is predictable from the independent variables. However, it has a major flaw:
* **It never decreases:** Every time you add a new feature or independent variable to your model, the standard $R^2$ will either stay the same or increase. 
* **The Overfitting Risk:** Even if you add completely random, irrelevant noise features (e.g., predicting house prices using the weather in another country), the standard $R^2$ can artificially increase, misleading you into believing the model is improving.

---

## How Adjusted $R^2$ Solves This

**Adjusted R-Squared penalizes the model for adding features that do not add predictive value.** * If you add a useful feature that significantly improves the model, Adjusted $R^2$ **increases**.
* If you add an irrelevant feature that adds no real value, Adjusted $R^2$ **decreases**.

---

## Mathematical Formula

The relationship between standard $R^2$ and Adjusted $R^2$ is defined as:

$$\text{Adjusted } R^2 = 1 - \left[ \frac{(1 - R^2)(n - 1)}{n - p - 1} \right]$$

Where:
* **$R^2$**: The standard Coefficient of Determination.
* **$n$**: The total number of observations (sample size/rows).
* **$p$**: The total number of independent variables (predictors/features).

### Understanding the Penalty Component:
As the number of predictors ($p$) increases, the denominator $(n - p - 1)$ becomes smaller. This increases the value of the fraction being subtracted from 1, pulling the overall Adjusted $R^2$ value down unless the standard $R^2$ increases enough to counteract the penalty.

---

## Key Differences Summary

| Feature | Standard $R^2$ | Adjusted $R^2$ |
| :--- | :--- | :--- |
| **Effect of adding more features** | Always increases or stays constant. | Increases *only* if the feature improves model quality; otherwise decreases. |
| **Reliability in Multiple Regression** | Low (prone to overestimating model fit). | High (provides a realistic look at feature utility). |
| **Comparative Use** | Best for Simple Linear Regression (1 feature). | Best for Multiple Linear Regression (>1 feature). |
| **Value Bounds** | Always between $0$ and $1$ (if a baseline intercept is included). | Can become negative if the model performs worse than a simple horizontal line mapping the mean. |

---

## Key Takeaways for Model Selection

1. **Compare Models Confidently:** When comparing two regression models with a different number of features, always use **Adjusted $R^2$** instead of standard $R^2$.
2. **Feature Pruning:** If your standard $R^2$ goes up slightly after adding a feature, but your Adjusted $R^2$ goes down, drop that feature—it is adding noise rather than signal.

```python
from sklearn.metrics import r2_score
```

## To know more about `r2_score` [visit](r2_score.ipynb)

```python
def adjusted_r2_score(y_true, y_pred, number_of_features):
    """
    Calculate adjusted R² score.

    Parameters
    ----------
    y_true : array-like
        Actual target values.
    y_pred : array-like
        Predicted target values.
    number_of_features : int
        Number of independent input features.

    Returns
    -------
    float
        Adjusted R² score.
    """
    n = len(y_true)
    p = number_of_features

    if n <= p + 1:
        raise ValueError(
            "Adjusted R² requires the number of observations "
            "to be greater than the number of features + 1."
        )

    r2 = r2_score(y_true, y_pred)

    return 1 - ((1 - r2) * (n - 1) / (n - p - 1))


 
```

```python
y_true = [45, 50, 55, 62, 65, 71, 78, 83, 85, 92]
y_pred = [44, 52, 54, 60, 67, 70, 76, 84, 86, 90]

score = adjusted_r2_score(
    y_true=y_true,
    y_pred=y_pred,
    number_of_features=2
)
print("Adjusted R²:", score)
```
