# Mean Squared Error (MSE) 

It is the intermediate step before calculating RMSE. It measures the average of the squared differences between your predicted and actual values.

**The Formula:** Mthematically, MSE removes the square root from the RMSE formula:
$$MSE=\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2$$


```python
def calculate_mse(actual, predicted):
    """
    Calculate Mean Squared Error using pure Python.
    """
    if len(actual) != len(predicted):
        raise ValueError("The lengths of actual and predicted lists must be equal.")
        
    n = len(actual)
    squared_errors = [(act - pred) ** 2 for act, pred in zip(actual, predicted)]
    
    return sum(squared_errors) / n

```

```python
def calculate_mse_v2(actual, predicted):
    if len(actual) != len(predicted):
        raise ValueError("The length of actual and predicted list must be equal")

    """
    Advatages is not creating array to store squared error , rather it will sum in same line 
    """
    squared_errors = 0.0

    for act , pred in zip(actual, predicted):
        squared_errors += pow((act-pred),2)

    return squared_errors/len(actual)
```

```python
# Example Usage:
actual_values = [20.5, 30.2, 28.1, 40.0]
predicted_values = [22.0, 29.5, 31.0, 38.5]

print("Custom MSE:", calculate_mse(actual_values, predicted_values))
print("Custom MSE-v2:", calculate_mse_v2(actual_values, predicted_values))
```
