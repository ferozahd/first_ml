**_What it's explain:_**

- Probabilites are should not mismatch
- Where a certain probability showing positive everytime this pobability or more than that should always positive 
- Same probability should not stay both positive and negative at same time
- If a certain probability showing negative every time this probability and less that that should always negative


__if any aforementioned arguments omitted, that definitely reduce the score__

````python
import numpy as np 
from sklearn.metrics import roc_auc_score
````

````python
def custom_roc_auc_score(y_true, y_proba):
    # Convert custom list to numpy array 
    y_true =np.asarray(y_true)
    y_score=np.asarray(y_proba)

    #Validation 
    if len(y_true) != len(y_proba) and len(y_true) !=0:
        raise ValueError("Y true and Y probabilities should have same length and not empty")

    if not np.all(np.isin(y_true,[0,1])) and not np.all((y_score >= 0) & (y_score <= 1)):
        raise ValueError("Y true value eithe should be 0 or 1")



    # Separate scores by actual class
    '''
    y_true = [1, 1, 0, 0]
    y_probability = [0.9, 0.7, 0.6, 0.2]
    '''
    #Positive socre will be 0.9,0.7 
    #Negative scores will be 0.6,0.2
    positive_scores = y_score[y_true == 1]
    negative_scores = y_score[y_true == 0] 

    
    # Correct ranking, probablility sequence is not messy , if 0.9 =1  , 0.7 =1 , 
    #Then 0.8 will be 1 , that will increase correct_rankings score by 1
    #But if 0.8=0 then tried_ranking will get one points
    correct_rankings = 0
    tied_rankings = 0
    
    # Compare every positive score with every negative score
    for positive_score in positive_scores:
        for negative_score in negative_scores:
            
            if positive_score > negative_score:
                correct_rankings += 1

            elif positive_score <= negative_score :
                tied_rankings += 1

    total_pairs = len(positive_scores) * len(negative_scores)

    auc = (
        correct_rankings   * tied_rankings
    ) / total_pairs

    return auc
    
````

````python
y_true = [1, 1, 0, 1]
y_probability = [0.9, 0.7, 0.6, 0.2]

score=custom_roc_auc_score(y_true, y_probability)
b_score=roc_auc_score(y_true, y_probability)
print(f'Custom method result {score}')
print(f'Biltin method result {b_score}')
````
