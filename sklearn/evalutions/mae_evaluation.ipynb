# MAE (Mean Absolute Error)

## The Core Components
MAE is the most intuitive regression evaluation metric. It measures the average absolute magnitude of the errors between your model's predictions and the true targets, giving all errors equal weight regardless of their direction.
- True Value ($y_i$): The actual ground-truth target variable for sample $i$, where $y_i \in \mathbb{R}$.
- Prediction ($\hat{y}_i$): The continuous scalar value estimated by your regression model for sample $i$.
- Absolute Loss (Absolute Error): The non-negative distance between the true value and the predicted value:

$$\vert{}y_i - \hat{y}_i\vert{}$$


## The Real Mathematical Formula
In rigorous statistical textbooks, MAE is defined as the arithmetic mean of the $L_1$ norms of the individual residual errors across a finite sample set.For a dataset containing $n$ samples, the formal mathematical definition is:

$$\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} \vert{}y_i - \hat{y}_i\vert{}$$


## Step-by-Step Walkthrough with Your Dataset
Let's evaluate the metric over your sample size ($n = 10$).

Actual values ($\mathbf{y}$): $[45, 50, 55, 62, 65, 71, 78, 83, 85, 92]$

Predicted values ($\mathbf{\hat{y}}$): $[45.09, 50.32, 55.54, 60.76, 65.99, 71.21, 76.44, 81.66, 86.88, 92.11]$

### Step A: Calculate individual absolute losses
We calculate the component$\vert{}y_i - \hat{y}_i\vert{}$ for each pair:

- $i=1: \vert{}45 - 45.09\vert{} = \vert{}-0.09\vert{} = \mathbf{0.09}$
- $i=2: \vert{}50 - 50.32\vert{} = \vert{}-0.32\vert{} = \mathbf{0.32}$
- $i=3: \vert{}55 - 55.54\vert{} = \vert{}-0.54\vert{} = \mathbf{0.54}$
- $i=4: \vert{}62 - 60.76\vert{} = \vert{}1.24\vert{} = \mathbf{1.24}$
- $i=5: \vert{}65 - 65.99\vert{} = \vert{}-0.99\vert{} = \mathbf{0.99}$
- $i=6: \vert{}71 - 71.21\vert{} = \vert{}-0.21\vert{} = \mathbf{0.21}$
- $i=7: \vert{}78 - 76.44\vert{} = \vert{}1.56\vert{} = \mathbf{1.56}$
- $i=8: \vert{}83 - 81.66\vert{} = \vert{}1.34\vert{} = \mathbf{1.34}$
- $i=9: \vert{}85 - 86.88\vert{} = \vert{}-1.88\vert{} = \mathbf{1.88}$
- $i=10:\vert{}92 - 92.11\vert{} = \vert{}-0.11\vert{} = \mathbf{0.11}$


### Step B: Sum the absolute losses

$$\sum_{i=1}^{10} \vert{}y_i - \hat{y}_i\vert{} = 0.09 + 0.32 + 0.54 + 1.24 + 0.99 + 0.21 + 1.56 + 1.34 + 1.88 + 0.11 = \mathbf{8.28}$$

### Step C: Divide by sample size ($n=10$)

$$\text{MAE} = \frac{8.28}{10} = \mathbf{0.828}$$

On average, your model's predictions miss the target value by approximately 0.828 units.


```python
import numpy as np
```

```python

def custom_mae(y_true, y_pred):
    """
    Custom implementation of Mean Absolute Error (MAE)
    mirroring the exact mathematical average of absolute losses.
    """
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    
    # Calculate absolute losses: |y - y_hat|
    absolute_losses = np.abs(y_true - y_pred)
    
    # Return the arithmetic mean
    return np.mean(absolute_losses)
```

```python

from sklearn.metrics import mean_absolute_error


# -------------------------------------------------------------
# Execution with your dataset
# -------------------------------------------------------------
actual_values = [45, 50, 55, 62, 65, 71, 78, 83, 85, 92]
predicted_values = [45.09, 50.32, 55.54, 60.76, 65.99, 71.21, 76.44, 81.66, 86.88, 92.11]

# 1. Built-in Scikit-Learn method
builtin_result = mean_absolute_error(actual_values, predicted_values)

# 2. Custom Method
custom_result = custom_mae(actual_values, predicted_values)

print(f"MAE (Scikit-Learn Built-in): {builtin_result:.3f}")
print(f"MAE (Custom Implementation):  {custom_result:.3f}")
```
