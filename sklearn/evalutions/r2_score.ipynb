# R-Squared ($R^2$) Score (Coefficient of Determination)

In regression analysis, the **R-Squared ($R^2$) score**, also known as the **Coefficient of Determination**, is a fundamental statistical metric used to evaluate the performance of a regression model. It measures the proportion of variance in the dependent variable (target) that can be predicted or explained by the independent variables (features).

---

## High-Level Concept

Think of $R^2$ as a relative comparison tool. Instead of telling you the raw error (like Mean Squared Error), it answers the question: 
> *"How much better is my model's prediction compared to simply guessing the average value of the target for every data point?"*



---

## Mathematical Formulation

The $R^2$ score is calculated using the following components:

### 1. Total Sum of Squares ($SS_{\text{tot}}$)
This represents the total variation in the actual data points ($y_i$) around their mean ($\bar{y}$). It is the error of a naive baseline model that always predicts the average value.
$$SS_{\text{tot}} = \sum_{i=1}^{n} (y_i - \bar{y})^2$$

### 2. Residual Sum of Squares ($SS_{\text{res}}$)
This represents the variation or discrepancy left unexplained after fitting your model. It is the sum of the squared differences between the actual values ($y_i$) and the predicted values ($\hat{y}_i$).
$$SS_{\text{res}} = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

### 3. $R^2$ Formula
Combining these two components:

$$R^2 = 1 - \frac{SS_{\text{res}}}{SS_{\text{tot}}}$$

---

## Interpreting the Values

The $R^2$ score typically falls within a scale of **0 to 1**, though it can mathematically fall below zero:

* **$R^2 = 1.0$ (Perfect Model):** The residual sum of squares ($SS_{\text{res}}$) is $0$. The model's predictions perfectly match the actual data points.
* **$R^2 = 0.0$ (Baseline Equivalent):** The model performs exactly as poorly as a simple horizontal line drawn at the average ($\bar{y}$) of the dataset.
* **$0 < R^2 < 1$ (Practical Performance):** An $R^2$ of $0.75$ means that **$75\%$** of the variance in the target variable is explained by the model, while the remaining $25\%$ is unexplained noise.
* **$R^2 < 0$ (Worse than the Mean):** If the model performs worse than simply predicting the average value for every observation, the ratio $\frac{SS_{\text{res}}}{SS_{\text{tot}}}$ becomes greater than 1, yielding a negative $R^2$.

---

## Pros and Cons of Standard $R^2$

### Pros
* **Scale-Invariant:** Since it is a ratio, it is dimensionless. You can use it to compare model fits across completely different scales and target units.
* **Highly Intuitive:** Translating model fit into an easily digestible percentage (e.g., "90% of the variance explained") is highly effective when communicating with stakeholders.

### Cons
* **Inflated by Feature Count:** Standard $R^2$ will monotonically increase (or stay the same) as you add more features to the model, regardless of whether those features carry genuine predictive signal.
* **Insensitive to Systematic Bias:** A model can have a high $R^2$ score while still producing systematically biased overpredictions or underpredictions if the data exhibits non-linear relationships that the model forces into a linear format.

# How the $R^2$ (R-squared) Score is Calculated

The $R^2$ (coefficient of determination) measures how well a model&apos;s 
predictions perform compared to a simple baseline
**the average (mean) value** 
of the actual target variables. 

Below is the step-by-step mathematical calculation using your specific dataset.

---

## 1. The Core Mathematical Formula

The formula to calculate $R^2$ is:

$$R^2 = 1 - \frac{SS_{\text{res}}}{SS_{\text{tot}}}$$

Where:
* **$SS_{\text{res}}$ (Sum of Squared Residuals):** The total squared error left over after our model makes its predictions.
* **$SS_{\text{tot}}$ (Total Sum of Squares):** The total baseline variation in the actual real-world data relative to its mean.

Conceptually, this translates to:
$$\text{R-squared} = 1 - \frac{\text{Unexplained Variation}}{\text{Total Variation}}$$

---

## 2. The Input Data

* **Actual values ($y$):** `[45, 50, 55, 62, 65, 71, 78, 83, 85, 92]` (Sample size $n = 10$)
* **Predicted values ($\hat{y}$):** `[45.09, 50.32, 55.54, 60.76, 65.99, 71.21, 76.44, 81.66, 86.88, 92.11]`

---

## 3. Step-by-Step Calculation

### Step 1: Find the Mean of the Actual Values ($\bar{y}$)
Add all the actual target values together and divide by the total number of samples ($n = 10$):

$$\bar{y} = \frac{45 + 50 + 55 + 62 + 65 + 71 + 78 + 83 + 85 + 92}{10} = \frac{686}{10} = \mathbf{68.6}$$

*Our baseline comparison mean score is **68.6**.*

---

### Step 2: Calculate the Total Sum of Squares ($SS_{\text{tot}}$)
This represents the total variance present in the actual data points before the model tries to explain any of it. For each actual value $y_i$, we calculate $(y_i - \bar{y})^2$ and sum them:

* $(45 - 68.6)^2 = (-23.6)^2 = 556.96$
* $(50 - 68.6)^2 = (-18.6)^2 = 345.96$
* $(55 - 68.6)^2 = (-13.6)^2 = 184.96$
* $(62 - 68.6)^2 = (-6.6)^2  = 43.56$
* $(65 - 68.6)^2 = (-3.6)^2  = 12.96$
* $(71 - 68.6)^2 = (2.4)^2    = 5.76$
* $(78 - 68.6)^2 = (9.4)^2    = 88.36$
* $(83 - 68.6)^2 = (14.4)^2   = 207.36$
* $(85 - 68.6)^2 = (16.4)^2   = 268.96$
* $(92 - 68.6)^2 = (23.4)^2   = 547.56$

$$SS_{\text{tot}} = 556.96 + 345.96 + 184.96 + 43.56 + 12.96 + 5.76 + 88.36 + 207.36 + 268.96 + 547.56 = \mathbf{2262.4}$$

---

### Step 3: Calculate the Sum of Squared Residuals ($SS_{\text{res}}$)
This measures the prediction error of our model (the distance between actual values $y_i$ and predicted values $\hat{y}_i$). For each pair, we calculate $(y_i - \hat{y}_i)^2$ and sum them:

* $(45 - 45.09)^2 = (-0.09)^2 = 0.0081$
* $(50 - 50.32)^2 = (-0.32)^2 = 0.1024$
* $(55 - 55.54)^2 = (-0.54)^2 = 0.2916$
* $(62 - 60.76)^2 = (1.24)^2  = 1.5376$
* $(65 - 65.99)^2 = (-0.99)^2 = 0.9801$
* $(71 - 71.21)^2 = (-0.21)^2 = 0.0441$
* $(78 - 76.44)^2 = (1.56)^2  = 2.4336$
* $(83 - 81.66)^2 = (1.34)^2  = 1.7956$
* $(85 - 86.88)^2 = (-1.88)^2 = 3.5344$
* $(92 - 92.11)^2 = (-0.11)^2 = 0.0121$

$$SS_{\text{res}} = 0.0081 + 0.1024 + 0.2916 + 1.5376 + 0.9801 + 0.0441 + 2.4336 + 1.7956 + 3.5344 + 0.0121 \approx \mathbf{10.74}$$

---

### Step 4: Plug the Totals into the $R^2$ Formula
Using our calculated values, we compute the final ratio:

$$R^2 = 1 - \frac{SS_{\text{res}}}{SS_{\text{tot}}}$$

$$R^2 = 1 - \frac{10.74}{2262.4}$$

$$R^2 \approx 1 - 0.004747$$

$$R^2 \approx \mathbf{0.9952}$$

---

## Conclusion
The model fails to explain only **$0.48\%$** of the variation in the dataset ($\frac{10.74}{2262.4} \approx 0.0048$). Subtracting this unexplained portion from $1$ (representing $100\%$) yields **$0.9952$**, meaning **$99.52\%$** of the variance in exam scores is successfully explained by study hours.
