# Root Mean Squared Error (RMSE) 

It is a standard metric used to evaluate regression models in machine learning. It measures the average distance between the predicted values and the actual values, expressed in the exact same units as the target variable. Lower RMSE values indicate a better fit and higher prediction accuracy.

 
**The Formula:** Mathematically, RMSE is the square root of the average of squared differences (errors) between the predicted and actual values:

$$RMSE=\sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2}$$

- n: Total number of data points
- $y_i$: Actual observed value
- $\hat{y}_{i}$: Value predicted by the model
- $\sum$: Summation of all data points from i = 1 to n


```python
import math
def calculate_rmse(actual, predicted):
    """
    Calculate Root Mean Squared Error using pure Python.
    """
    if len(actual) != len(predicted):
        raise ValueError("The lengths of actual and predicted lists must be equal.")
        
    n = len(actual)
    squared_errors = [(act - pred) ** 2 for act, pred in zip(actual, predicted)]
    mean_squared_error = sum(squared_errors) / n
    
    return math.sqrt(mean_squared_error)
```

>The value "1.83" shares the exact same unit as your target variable. If your model predicts house prices in thousands of dollars, your predictions are off by an average of $1,830. If it predicts temperature in Celsius, it is off by 1.83°C.

```python
actual_values = [20.5, 30.2, 28.1, 40.0]
predicted_values = [22.0, 29.5, 31.0, 38.5]

print("Custom RMSE:", calculate_rmse(actual_values, predicted_values))
```
