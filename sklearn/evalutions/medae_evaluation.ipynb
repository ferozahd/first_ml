# Median Absolute Error
It is a metric used to evaluate the performance of regression models (models that predict continuous numbers, like housing prices or stock values) by measuring the average distance between the model's predictions and the actual true values.

**Formula:**
$$\text{MedAE} = \arg\min_{m \in \mathbb{R}} \sum_{i=1}^{n} \big\vert{} \vert{}y_i - \hat{y}_i\vert{} - m \big\vert{}$$

**Simplifying formula:**

**Example:**
Step 1: Calculate the Absolute Losses

($e_i = \vert{}y_i - \hat{y}_i\vert{}$)We calculate the absolute difference for all 10 pairs:
- $e_1 = \vert{}45 - 45.09\vert{} = \vert{}-0.09\vert{} = \mathbf{0.09}$
- $e_2 = \vert{}50 - 50.32\vert{} = \vert{}-0.32\vert{} = \mathbf{0.32}$
- $e_3 = \vert{}55 - 55.54\vert{} = \vert{}-0.54\vert{} = \mathbf{0.54}$
- $e_4 = \vert{}62 - 60.76\vert{} = \vert{}1.24\vert{} = \mathbf{1.24}$
- $e_5 = \vert{}65 - 65.99\vert{} = \vert{}-0.99\vert{} = \mathbf{0.99}$
- $e_6 = \vert{}71 - 71.21\vert{} = \vert{}-0.21\vert{} = \mathbf{0.21}$
- $e_7 = \vert{}78 - 76.44\vert{} = \vert{}1.56\vert{} = \mathbf{1.56}$
- $e_8 = \vert{}83 - 81.66\vert{} = \vert{}1.34\vert{} = \mathbf{1.34}$
- $e_9 = \vert{}85 - 86.88\vert{} = \vert{}-1.88\vert{} = \mathbf{1.88}$
- $e_{10} = \vert{}92 - 92.11\vert{} = \vert{}-0.11\vert{} = \mathbf{0.11}$

Our unsorted absolute loss set is:

$$E = \{0.09, 0.32, 0.54, 1.24, 0.99, 0.21, 1.56, 1.34, 1.88, 0.11\}$$

Step 2: Short in accending order

$$\{0.09, 0.11, 0.21, 0.32, \mathbf{0.54}, \mathbf{0.99}, 1.24, 1.34, 1.56, 1.88\}$$


Step 3: Evaluate the Median for Even Cardinality ($n = 10$)

Because $n = 10$ is an even number, we must find the arithmetic mean of the two central elements at indices $\frac{n}{2}$ and $\frac{n}{2} + 1$ (the 5th and 6th elements in our sorted list):
- 5th element ($e_{\pi(5)}$) = $0.54$
- 6th element ($e_{\pi(6)}$) = $0.99$

$$\text{MedAE} = \frac{e_{\pi(5)} + e_{\pi(6)}}{2}$$

$$\text{MedAE} = \frac{0.54 + 0.99}{2} = \frac{1.53}{2} = \mathbf{0.765}$$


```python

def custom_medae(y_true, y_pred):
    """
    Custom implementation of Median Absolute Error (MedAE)
    mirroring the exact mathematical formulation using order statistics.
    """
    # Convert inputs to numpy arrays for element-wise operations
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    
    # Step 1: Compute absolute losses |y_i - y_hat_i|
    absolute_losses = np.abs(y_true - y_pred)
    
    # Step 2: Sort the losses to establish order statistics
    sorted_losses = np.sort(absolute_losses)
    
    n = len(sorted_losses)
    
    # Step 3: Evaluate based on cardinality (odd vs even)
    if n % 2 != 0:
        # If n is odd, pick the exact middle element (index: (n-1)//2)
        medae_val = sorted_losses[n // 2]
    else:
        # If n is even, average the two central elements
        low_middle = sorted_losses[(n // 2) - 1]
        high_middle = sorted_losses[n // 2]
        medae_val = (low_middle + high_middle) / 2.0
        
    return medae_val
```

```python
import numpy as np
from sklearn.metrics import median_absolute_error

# -------------------------------------------------------------
# Execution with your dataset
# -------------------------------------------------------------

# Your precise sample data (n = 10)
actual_values = [45, 50, 55, 62, 65, 71, 78, 83, 85, 92]
predicted_values = [45.09, 50.32, 55.54, 60.76, 65.99, 71.21, 76.44, 81.66, 86.88, 92.11]

# 1. Built-in Method (scikit-learn)
builtin_result = median_absolute_error(actual_values, predicted_values)

# 2. Custom Method
custom_result = custom_medae(actual_values, predicted_values)

# Display Results
print(f"Dataset Size (n): {len(actual_values)}")
print(f"MedAE (Built-in scikit-learn): {builtin_result:.3f}")
print(f"MedAE (Custom implementation): {custom_result:.3f}")

# Verification check
assert np.isclose(builtin_result, custom_result), "Calculations do not match!"
print("\nSuccess: Both methods yield identical results.")
```
