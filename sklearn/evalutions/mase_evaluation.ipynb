# Mean Absolute Scaled Error

MASE is primarily used in time series forecasting. It evaluates a model by comparing its absolute losses against the absolute losses of a baseline Naïve Forecast (a model that simply predicts the next value will be identical to the current value).

**Formula:**
$$\text{MASE} = \frac{\displaystyle\frac{1}{m} \sum_{t=n+1}^{n+m} \vert{}y_t - \hat{y}_t\vert{}}{\displaystyle\frac{1}{n-1} \sum_{i=2}^{n} \vert{}y_i - y_{i-1}\vert{}}$$


## Step-by-Step Walkthrough with Your Dataset
Let’s assume your data represents a sequential time series. To keep things clean, we will evaluate the model in-sample over your dataset ($n = 10$).

Actual values ($\mathbf{y}$): $[45, 50, 55, 62, 65, 71, 78, 83, 85, 92]$

Predicted values ($\mathbf{\hat{y}}$): $[45.09, 50.32, 55.54, 60.76, 65.99, 71.21, 76.44, 81.66, 86.88, 92.11]$

### Step 1:  Calculate the Model's Absolute Losses and Mean ($t = 1$ to $10$)
$$\sum_{t=1}^{10} \vert{}y_t - \hat{y}_t\vert{} = 0.09 + 0.32 + 0.54 + 1.24 + 0.99 + 0.21 + 1.56 + 1.34 + 1.88 + 0.11 = 8.28$$


   $$\text{MAE}_{\text{model}} = \frac{8.28}{10} = \mathbf{0.828}$$

### Calculate the base line Naïve Absolute Losses ($i = 2$ to $10$)

We find the absolute differences between consecutive historical actual values:
- $\vert{}y_2 - y_1\vert{} = \vert{}50 - 45\vert{} = 5$
- $\vert{}y_3 - y_2\vert{} = \vert{}55 - 50\vert{} = 5$
- $\vert{}y_4 - y_3\vert{} = \vert{}62 - 55\vert{} = 7$
- $\vert{}y_5 - y_4\vert{} = \vert{}65 - 62\vert{} = 3$
- $\vert{}y_6 - y_5\vert{} = \vert{}71 - 65\vert{} = 6$
- $\vert{}y_7 - y_6\vert{} = \vert{}78 - 71\vert{} = 7$
- $\vert{}y_8 - y_7\vert{} = \vert{}83 - 78\vert{} = 5$
- $\vert{}y_9 - y_8\vert{} = \vert{}85 - 83\vert{} = 2$
- $\vert{}y_{10} - y_9\vert{} = \vert{}92 - 85\vert{} = 7$

$$\text{Sum of Naïve Losses} = 5 + 5 + 7 + 3 + 6 + 7 + 5 + 2 + 7 = 47$$

$$\text{MAE}_{\text{baseline}} = \frac{47}{10 - 1} = \frac{47}{9} \approx \mathbf{5.2222}$$

### Compute the Final Ratio

$$\text{MASE} = \frac{\text{MAE}_{\text{model}}}{\text{MAE}_{\text{baseline}}} = \frac{0.828}{5.2222} \approx \mathbf{0.1585}$$

```python
import numpy as np
```

```python
def custom_mase(y_train, y_true, y_pred):
    """
    Custom implementation of Mean Absolute Scaled Error (MASE)
    mirroring the exact mathematical ratio formulation.
    """
    y_train = np.array(y_train)
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    
    # 1. Numerator: MAE of the model evaluations
    mae_model = np.mean(np.abs(y_true - y_pred))
    
    # 2. Denominator: MAE of the baseline naive forecast on training data
    naive_absolute_losses = np.abs(y_train[1:] - y_train[:-1])
    mae_baseline = np.mean(naive_absolute_losses)
    
    # 3. Ratio
    return mae_model / mae_baseline
```

```python

# Requires: pip install sktime
#from sktime.performance_metrics.forecasting import mean_absolute_scaled_error


# -------------------------------------------------------------
# Execution with your dataset (Evaluating In-Sample)
# -------------------------------------------------------------
actual_values = [45, 50, 55, 62, 65, 71, 78, 83, 85, 92]
predicted_values = [45.09, 50.32, 55.54, 60.76, 65.99, 71.21, 76.44, 81.66, 86.88, 92.11]

# Custom Method
custom_result = custom_mase(
    y_train=actual_values, 
    y_true=actual_values, 
    y_pred=predicted_values
)

# Built-in Method (sktime)
# Note: sktime expects 2D arrays or pandas Series natively, or matching array structures
'''
builtin_result = mean_absolute_scaled_error(
    y_true=np.array(actual_values), 
    y_pred=np.array(predicted_values), 
    y_train=np.array(actual_values)
)'''

print(f"MASE (Custom implementation): {custom_result:.4f}")
#print(f"MASE (sktime library):        {builtin_result:.4f}")
```
