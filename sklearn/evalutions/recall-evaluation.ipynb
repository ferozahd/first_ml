# Exploring `recall`
**Formula:**
$$\text{Recall} = \frac{\text{True positive}}{\text{True Positive + False Negative}}$$

Why do we need?

```text 
let' see a question frist : 
A bank processes two million transcation daily, but only 0.2% are fraudulent. How would you build and evalute a fraud-detection system ?

problem is legitimate almost 100% (99.8) and fraudulent nearly 0$ (0.2)
```
problem here : 

- A good model can be overfit easily
- Model almost predict every transaction is legitimate
- We need something that focus on small part as big picture
- We have 4000 fraud and 1996000 is legitimate but I want to focus only on 4000
**Recall solves this problem**

1. It will ignore the huge legitimate majority problem
2. It directly measure missed fraud , ignore legitimate competely
3. It convert percentage into operational impact
```text 
Explain number of fraud cauth and number of fraud are missed 
```
5. It's allow model to be compared fairly
```text 
It's focusing 2 differents model recall score compare 
```
6. It helps choose the fraud detection threshold
```text 
Not directly providing threshold butoffer precession score so developer can choose good threshold 
```
7. It can monitor model direction 

````python
from sklearn.metrics import recall_score

y_true = [1, 1, 1, 1, 0]
y_pred = [1, 1, 1, 0, 0]

recall = recall_score(y_true, y_pred)

print(recall)  # 0.75
````
