```python
import numpy as np
from sklearn.metrics import average_precision_score
```

```python


def custom_average_precision(y_true, y_scores):
    """
    Custom implementation of Average Precision (AP)
    mirroring the rank-ordered precision summation formula.
    """
    y_true = np.array(y_true)
    y_scores = np.array(y_scores)
    
    # Sort indices by prediction scores in descending order
    desc_indices = np.argsort(y_scores)[::-1]
    y_true_sorted = y_true[desc_indices]
    
    # Calculate running True Positives and total item positions
    running_tps = np.cumsum(y_true_sorted)
    positions = np.arange(1, len(y_true_sorted) + 1)
    
    # Precision at each rank step: running_tps / running_total
    precisions = running_tps / positions
    
    # Sum precisions only at positions where the true item is positive (y == 1)
    total_positives = np.sum(y_true)
    ap_score = np.sum(precisions * y_true_sorted) / total_positives
    
    return ap_score
```

# Execution with our tracked walkthrough data

```python
true_labels = [1, 0, 1, 0, 0]
model_probabilities = [0.95, 0.80, 0.65, 0.40, 0.15]

# 1. Built-in Scikit-Learn method
builtin_result = average_precision_score(true_labels, model_probabilities)

# 2. Custom Method
custom_result = custom_average_precision(true_labels, model_probabilities)

print(f"Average Precision (Scikit-Learn Built-in): {builtin_result:.4f}")
print(f"Average Precision (Custom Implementation):  {custom_result:.4f}")
```
