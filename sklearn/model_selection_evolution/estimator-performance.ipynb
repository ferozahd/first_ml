## Estimator performance 
- Loading sample data from sklean 
- Checking input and output shape 

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn import datasets
from sklearn import svm

x, y = datasets.load_iris(return_X_y=True)
print (f'X shape is : ',x.shape)
print(f'X shape means X is an 2d array with rows {x.shape[1]} and length of array is {x.shape[0]}')
print (f'Y shape is : ',y.shape)
print(f'Y shape means Y is 1D array, length of array is {y.shape[0]}')
```

```python
x_train, x_test, y_train, y_test = train_test_split(
     x, y, test_size=0.4, random_state=0)

#Spliting data in 2 parts, first part to train the model and 2nd part to test the model accuracy 
print(f'Number of traning X len : {x_train.shape[0]} and y len : {y_train.shape[0]}')
print(f'Number of testing X len : {x_test.shape[0]} and y len : {y_test.shape[0]}')

clf = svm.SVC(kernel='linear', C=1).fit(x_train, y_train)
clf.score(x_test, y_test)
```

### Checking cross validation score 
- cross_val_score useally train test split internally 
  - First parameter is model, which model should use to train and test 
  - 2nd parameter Data input sample 
  - 3rd parameter Data output sample 
  - 4th cv number of spliting 
    - number by default 
    - KFold 
    - ShuffleSplit
    - RepeatedKFold
  - 5th scoring evaluate matric 
    - accuracy 
    - precision 
    - recall 
    - f1
    - f1_macro 
    - r2
    - neg_mean_squared_error

```python
from sklearn.model_selection import cross_val_score
clf = svm.SVC(kernel='linear', C=1, random_state=42)
scores = cross_val_score(clf, x, y, cv=5,scoring="accuracy")
print(scores)
```
