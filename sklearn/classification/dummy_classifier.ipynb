# Dummy Classifier

## Introduction

A `DummyClassifier` is a simple baseline classification model.

It does not learn relationships between input features and the target. Instead, it makes predictions using a simple rule, such as predicting the most frequent class.

The main purpose of a Dummy Classifier is to compare its performance with real machine learning models.

If a real model cannot perform better than the Dummy Classifier, the real model may not be useful.

## Objective

In this notebook, we will:

1. Create a simple classification dataset.
2. Split the dataset into training and testing sets.
3. Train a Dummy Classifier.
4. Make predictions.
5. Evaluate the model.

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
# Create a simple binary classification dataset

X, y = make_classification(
    n_samples=500,
    n_features=5,
    n_informative=3,
    n_redundant=0,
    random_state=42
)

print("Feature shape:", X.shape)
print("Target shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y )

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
dummy_model = DummyClassifier(strategy="most_frequent")
```

## Model Strategy

The `most_frequent` strategy always predicts the class that appears most frequently in the training dataset.

For example, if class `0` is the most common class, the model will predict class `0` for every test sample.

```python
dummy_model.fit(X_train, y_train)
```

```python
y_pred = dummy_model.predict(X_test)

print("First 10 predictions:", y_pred[:10])
print("First 10 actual values:", y_test[:10])
```

```python
accuracy = accuracy_score(y_test, y_pred)

print(f"Dummy Classifier Accuracy: {accuracy:.2f}")
```

```python
print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)
```

```python
confusion_matrix_result = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(confusion_matrix_result)
```

```python
import pandas as pd

results = pd.DataFrame({
    "Actual": y_test,
    "Predicted": y_pred
})

results.head(10)
```

## Conclusion

The Dummy Classifier provides a baseline accuracy for the classification problem.

Because the `most_frequent` strategy predicts only the most common class, it does not identify meaningful relationships in the dataset.

A real classification model, such as Logistic Regression, Decision Tree, Random Forest, KNN or SVC, should perform better than this baseline.

If a real model performs similarly to or worse than the Dummy Classifier, the dataset, features or model configuration should be investigated.
